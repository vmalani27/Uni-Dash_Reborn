```python

sample_plan = {
    "Institutional Sender": 70,
    "External / Misc": 40,
    "External Academic Platform": 25,
    "Student / Peer": 15,
}
```

```python

samples = []

for source, n in sample_plan.items():
    subset = df[df["label_source"] == source]
    samples.append(subset.sample(n=n, random_state=42))

eval_df = pd.concat(samples).sample(frac=1, random_state=42).reset_index(drop=True)

eval_df.to_csv("llm_eval_subset_v1.csv", index=False)
```
